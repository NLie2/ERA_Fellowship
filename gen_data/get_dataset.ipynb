```python
import pandas as pd 
from datasets import load_dataset
```

```python
from pathlib import Path
import os
# Get the current working directory (where the notebook is)
notebook_dir = Path.cwd()

# Move up to the root directory (parent of 'notebooks')
project_root = notebook_dir.parent
datasets_path = project_root / "datasets"

# Ensure the datasets folder exists (create if it doesn't)
datasets_path.mkdir(parents=True, exist_ok=True)
```

```python
# df = pd.DataFrame(load_dataset("sevdeawesome/jailbreak_success")["train"])
# df.to_pickle(datasets_path/"jailbreak_success.pkl")


df = pd.DataFrame(load_dataset('JailbreakBench/JBB-Behaviors', 'judge_comparison')["test"])
df = df[(df["human1"] == 1) & (df["human2"] == 1) & (df["human3"] == 1)]
# # filter for rows that have "adversarial" and "harmful" in data_type

# # sample 100 random 

len(df)
```

```python
# rename "goal" to "original_prompt"
df = df.rename(columns={"goal": "original_prompt"})
df = df.rename(columns={"prompt": "jailbreak_prompt"})

df = df[["original_prompt", "jailbreak_prompt"]]
# drop the original "prompt" column

# save in validation data 
df.to_csv(f"/data/nathalie_maria_kirch/ERA_Fellowship/datasets/validation/jailbreak_bench.csv")
```
